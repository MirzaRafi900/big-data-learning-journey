**Create Dataset**

```python
data = [
    ("Kuala Lumpur", 1000),
    ("Kuala Lumpur", 500),
    ("Selangor", 700),
    ("Selangor", 300),
    ("Penang", 500),
    ("Penang", 400)
]

df = spark.createDataFrame(data, ["City","Sales"])

display(df)
```

**Showing Data**

```python
df.show()
```

**Showing Column Type**

```python
df.printSchema()
```

**Selecting Column**

```python
df.select("City").show()
```

**Filtering Row**

```python
df.filter(df["Sales"] > 500).show()
```

**Aggregating Data**

```python
from pyspark.sql.functions import sum, avg, max, min

df.groupBy("City").agg(

    sum("Sales").alias("Total Sales"),
    avg("Sales").alias("Average Sales"),
    max("Sales").alias("Highest Sales"),
    min("Sales").alias("Lowest Sales")

).show()
```
