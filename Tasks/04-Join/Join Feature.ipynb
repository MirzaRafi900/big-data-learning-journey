## **Creating Dataset**

```python
customer = [
    (1, "Ali"),
    (2, "Sara"),
    (3, "John")
]

customer_df = spark.createDataFrame(customer, ["Id", "Name"])
customer_df.show()
```

```python
orders = [
    (1, "Laptop"),
    (2, "Phone"),
    (3, "Mouse")
]

orders_df = spark.createDataFrame(orders, ["Id", "Product"])

orders_df.show()
```

```python
sales = [
    (1, 1000),
    (1, 500),
    (2, 1500),
    (2, 1000),
    (3, 3000),
    (3, 500)
]

sales_df = spark.createDataFrame(
    sales,
    ["ID", "Sales"]
)

sales_df.show()
```

#**Inner Join**

```python
join_df = customer_df.join(
    orders_df,
    on = "ID",
    how = "inner"
)

join_df.show()
```

# **Right Join**

```python
sale_amount_df = join_df.join(
    sales_df,
    on = "ID",
    how = "right"
)

sale_amount_df.show()
```

# **Aggregation**

```python
from pyspark.sql.functions import sum, avg, max, min

sale_amount_df.groupBy("Name").agg(
    sum("Sales").alias("Total Sales"),
    avg("Sales").alias("Average Sales"),
    max("Sales").alias("Highest Sales"),
    min("Sales").alias("Lowest Sales")
).show()
```
