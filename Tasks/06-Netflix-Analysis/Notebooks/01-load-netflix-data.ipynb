## **EDA**

### **Movie Counts**

```python
df.groupBy("type").count().show()
```

### **Top 10 Countries**

```python
df.groupBy("country").count().orderBy(desc("Count")).show(30)
```

### **Missing Values**

```python
df.select(
    [
        count(
            when(
                col(c).isNull(),c
            )
        ).alias(c)
        for c in df.columns
    ]
).show()
```
