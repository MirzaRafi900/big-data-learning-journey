### **Missing Values**

```python
df.select(
    [
        count(
            when(
                col(c).isNull(),c
            )
        ).alias(c)
        for c in df.columns
    ]
).show()
```

**Here Missing Values are,**
**director - 2636,**
**Cast - 826,**
**Countries - 832**

**Missing Value Solve**

```python
df_clean = df.fillna(
                        {
                            "director": "Unknown",
                            "cast" : "Unknown",
                            "country" : "Unknown"
                         }
                    )
df_clean.show(5)
```

```python
df_clean.select([
    count(
        when(col(c).isNull(),c)
    ).alias(c)
    for c in df_clean.columns
]).show()
```
